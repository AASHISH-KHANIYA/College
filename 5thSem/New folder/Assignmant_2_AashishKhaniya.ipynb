### Task 3:   
1. On the dataframe generated above, please answer the following:
    
    1. Extract Data of only India and make into a new dataframe.<br>
    2. Extract only the total cases column.<br>
    3. Convert total cases column into percentage. Percentage of total cases. Total cases is the number of cases as on the last date of the dataset. Use Numpy. DO NOT WRITE A LOOP!<br>
    4. Add this newly generated column to the India Dataframe.
-----------------------------------------------------------------------------------------------------------------

```python
# 1. Extract India data into a new dataframe
india_df = df[df['location'] == 'India'].copy()
india_df = india_df.sort_values('date').reset_index(drop=True)

# 2. Extract only the total cases column
total_cases_col = india_df['total_cases']

# 3. Convert total cases to percentage (relative to last date) using NumPy — no loop
last_date = india_df['date'].max()
total_cases_on_last_date = india_df.loc[india_df['date'] == last_date, 'total_cases'].values[0]
total_cases_percentage = (np.array(total_cases_col) / total_cases_on_last_date) * 100

# 4. Add the new column to the India dataframe
india_df['total_cases_percentage'] = total_cases_percentage

print(f"Total cases on last date ({last_date.date()}): {total_cases_on_last_date:,.0f}")
india_df[['date', 'total_cases', 'total_cases_percentage']].head(10)
```

### Task 4:   
1. On the dataframe generated above, please answer the following:
    
    1. extract the total deaths and total cases as a numpy array<br>
    2. transpose the array<br>
    3. crate new array which contains only total cases
    4. from the total cases row , extract only rows which are more than 10000
-----------------------------------------------------------------------------------------------------------------

```python
# 1. Extract total deaths and total cases as a NumPy array
cases_deaths_array = india_df[['total_deaths', 'total_cases']].to_numpy()
print("Original array shape:", cases_deaths_array.shape)
print(cases_deaths_array[:5])

# 2. Transpose the array
transposed_array = cases_deaths_array.T
print("\nTransposed array shape:", transposed_array.shape)
print(transposed_array)

# 3. Create a new array containing only total cases
total_cases_array = transposed_array[1]
print("\nTotal cases array:")
print(total_cases_array)

# 4. From the total cases row, extract only values greater than 10000
cases_above_10000 = total_cases_array[total_cases_array > 10000]
print(f"\nTotal cases values > 10,000: {len(cases_above_10000)} entries")
print(cases_above_10000)
```

### Task 5:   
1. On the dataframe generated above, please answer the following:
    
    1. Extract data only for 18-08-2020.<br>
    2. which country has the max and min
    3. take the top 10 and bottom 10 of total cases and total deaths
-----------------------------------------------------------------------------------------------------------------

```python
# 1. Extract data only for 18-08-2020
target_date = pd.to_datetime('2020-08-18')
df_aug18 = df[df['date'] == target_date].copy()
print(f"Rows on {target_date.date()}: {len(df_aug18)}")
df_aug18.head()

# 2. Country with max and min total cases (excluding aggregates with missing continent)
country_data = df_aug18[df_aug18['continent'].notna()]

max_cases_row = country_data.loc[country_data['total_cases'].idxmax()]
min_cases_row = country_data.loc[country_data['total_cases'].idxmin()]
max_deaths_row = country_data.loc[country_data['total_deaths'].idxmax()]
min_deaths_row = country_data.loc[country_data['total_deaths'].idxmin()]

print(f"\nMax total cases: {max_cases_row['location']} ({max_cases_row['total_cases']:,.0f})")
print(f"Min total cases: {min_cases_row['location']} ({min_cases_row['total_cases']:,.0f})")
print(f"Max total deaths: {max_deaths_row['location']} ({max_deaths_row['total_deaths']:,.0f})")
print(f"Min total deaths: {min_deaths_row['location']} ({min_deaths_row['total_deaths']:,.0f})")

# 3. Top 10 and bottom 10 of total cases and total deaths
print("\n--- Top 10 Total Cases ---")
print(country_data.nlargest(10, 'total_cases')[['location', 'continent', 'total_cases', 'total_deaths']])

print("\n--- Bottom 10 Total Cases ---")
print(country_data.nsmallest(10, 'total_cases')[['location', 'continent', 'total_cases', 'total_deaths']])

print("\n--- Top 10 Total Deaths ---")
print(country_data.nlargest(10, 'total_deaths')[['location', 'continent', 'total_cases', 'total_deaths']])

print("\n--- Bottom 10 Total Deaths ---")
print(country_data.nsmallest(10, 'total_deaths')[['location', 'continent', 'total_cases', 'total_deaths']])
```

### Task 6:  
1. On the dataframe generated above, please answer the following:
    
    1. Use groupby and<br>
        a. print number of countries in each group<br>
        b. get the total cases and deaths count for each continent as on 18-08-2020.<br>
        
    2. use apply() and <br>
        a. calculate the mortality rate for each row. and add that mortality data as a new column to the dataframe<br>
        
    3. do the same without apply<br>
    
-----------------------------------------------------------------------------------------------------------------

```python
# 1. Groupby operations
# a. Number of countries in each continent
countries_per_continent = df.groupby('continent')['location'].nunique()
print("Number of countries in each continent:")
print(countries_per_continent)

# b. Total cases and deaths for each continent on 18-08-2020
continent_summary = df_aug18.groupby('continent').agg(
    total_cases=('total_cases', 'sum'),
    total_deaths=('total_deaths', 'sum')
)
print("\nContinent-wise totals on 18-08-2020:")
print(continent_summary)

# 2. Mortality rate using apply()
def calc_mortality_rate(row):
    if pd.notna(row['total_cases']) and row['total_cases'] > 0:
        return row['total_deaths'] / row['total_cases']
    return np.nan

df_aug18['mortality_rate_apply'] = df_aug18.apply(calc_mortality_rate, axis=1)
print("\nMortality rate (using apply) - sample:")
print(df_aug18[['location', 'total_cases', 'total_deaths', 'mortality_rate_apply']].head())

# 3. Same calculation without apply()
df_aug18['mortality_rate_no_apply'] = df_aug18['total_deaths'] / df_aug18['total_cases']
print("\nMortality rate (without apply) - sample:")
print(df_aug18[['location', 'total_cases', 'total_deaths', 'mortality_rate_no_apply']].head())
```
