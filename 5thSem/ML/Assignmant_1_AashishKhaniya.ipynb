**Exercise 6:** Write a NumPy program to create an element-wise comparison (greater, greater_equal, less and less_equal) of two given arrays

```python
def compare_arrays(arr1, arr2):
    if not isinstance(arr1, np.ndarray) or not isinstance(arr2, np.ndarray):
        raise TypeError("Both inputs must be NumPy arrays.")
    if arr1.shape != arr2.shape:
        raise ValueError("Arrays must have the same shape for element-wise comparison.")
    return {
        "greater": np.greater(arr1, arr2),
        "greater_equal": np.greater_equal(arr1, arr2),
        "less": np.less(arr1, arr2),
        "less_equal": np.less_equal(arr1, arr2)
    }
if __name__ == "__main__":
    array1 = np.array([3, 5, 7, 9])
    array2 = np.array([2, 5, 8, 9])
    try:
        results = compare_arrays(array1, array2)
        print("Array 1:", array1)
        print("Array 2:", array2)
        print("\nElement-wise Comparisons:")
        for key, value in results.items():
            print(f"{key}: {value}")
    except (TypeError, ValueError) as e:
        print("Error:", e)
```

**Exercise 10:** Write a program in python to display prime numbers from x to y (here x and y are user given values).

```python
x,y=map(int,input().split())
for n in range(x,y+1):
    if n>1:
        for i in range(2,int(n**0.5)+1):
            if n%i==0:
                break
        else:
            print(n,end=' ')
```
